```python
import sys
import subprocess
from pathlib import Path

venv_python = Path.cwd() / ".venv" / "Scripts" / "python.exe"

if not venv_python.exists():
    raise FileNotFoundError(
        f"Project Python was not found at {venv_python}. "
        "Run this notebook from the OCR folder."
    )

print(f"Notebook kernel: {sys.executable}")
print(f"Installing with: {venv_python}")

packages = [
    "paddlepaddle==3.0.0",
    "paddleocr==2.9.1",
    "pdf2image"
]

for package in packages:
    subprocess.check_call([
        str(venv_python), "-m", "pip", "install", package
    ])

print("Python Paddle dependencies installed successfully.")
```
